# Retrieval analysis
Compare saved benchmark reports. Run the benchmark script first. Partial-corpus runs are smoke checks, not leaderboard scores.

```python
import json
from pathlib import Path

root = Path.cwd()
if root.name == "notebooks":
    root = root.parent
reports = [
    (p, json.loads(p.read_text(encoding="utf-8")))
    for p in sorted((root / "evals/results").glob("*.json"))
]
print(f"Found {len(reports)} reports")
```

```python
for path, report in reports:
    print(path.name, report["mode"], report["text_backend"])
    print(
        "Partial corpus:",
        report["dataset"]["partial_corpus"],
        "Queries:",
        report["evaluated_queries"],
    )
    print("Metrics:", report["metrics"])
    print("P95 retrieval seconds:", report["latency_p95_seconds"])
```

```python
for path, report in reports:
    metric = "ndcg@" + str(report["k"])
    worst = sorted(report["queries"], key=lambda row: row[metric])[:5]
    print(path.name, "Lowest-scoring query IDs:", [(r["query_id"], r[metric]) for r in worst])
```
